```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Cargar MNIST
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

# Preprocesamiento
x_train = x_train.reshape(-1, 784) / 255.0
x_test  = x_test.reshape(-1, 784) / 255.0

# One-hot
y_train = keras.utils.to_categorical(y_train, 10)
y_test  = keras.utils.to_categorical(y_test, 10)

# Construcción de la red
model = keras.Sequential([
    layers.Input(shape=(784,)),
    layers.Dense(30, activation="sigmoid"),
    layers.Dense(10, activation="sigmoid")
])

# SGD con tasa 0.5 igual al original
optimizer = keras.optimizers.Adam(learning_rate=0.001)

# Costo cuadrático para replicar exactamente tu código original
model.compile(
    optimizer=optimizer,
    loss="mean_squared_error",
    metrics=["accuracy"]
)

# Entrenamiento equivalente (15 épocas, batch 10)
model.fit(
    x_train, y_train,
    epochs=15,
    batch_size=10,
    validation_data=(x_test, y_test)
)

# Guardar el modelo
model.save("red_prueba.keras")
print("Entrenamiento terminado y red guardada en red_prueba.keras")
```
